```python

import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
# Get the shape of the matrix.
data.shape
data.dtype
```

```python
# Find the avg age of student.
age = data[:,:1]
print(f"Avg Age is {age.mean()}.")
```

```python
# Extract Maths marks of all studentd.
Math_marks = data[:,1:2]
print(f"Maths Marks are {Math_marks}.")
```

```python
# Find the Highest Science Marks
Science_marks = data[:,2:]
print(f"Maximum Marks is {Science_marks.max()}.")
```

```python
More_then_90 = (Math_marks > 90)
More_then_90
```

```python
# Increse Math marks of all Students by 5.
Incresed_math_Marks = (Math_marks + 5)
Incresed_math_Marks
```

```python
# Find how many students are younger than 19.
More_then_19 = (age > 19)
More_then_19
```

```python
print(f"Avg Math marks {Math_marks.mean()}.")
print(f"Avg Science Marks {Science_marks.mean()}.")
```
