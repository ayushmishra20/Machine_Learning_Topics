#### Feature Extraction

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(r'anime.csv')
df.head()
```

```python
df.loc[4]['Title']
```

```python
# Extract Episodes
def Extract_episodes(txt):
    check = False
    data = ""
    for i in txt:
        if i == ")":
            check = False
            return data
        if check == True:
            data = data +i
        if i == '(':
            check = True
```

```python
#make a new column for episode count
df['Episodes'] = df['Title'].apply(Extract_episodes)
df.head()
```

```python
df['Episodes'] = df['Episodes'].str.replace(" eps", "")
df.head()
```

```python
df['Episodes'] = df['Episodes'].astype(int)
df.head()
```

```python
#make a new column for time stamp
df.loc[3]['Title']
```

```python
def Extract_time(txt):
    check = False 
    data = ""
    for i in range(len(txt)):
        if txt[i] == ')':
            for j in range(i+1, i+20):
                data += txt[j]
            return data

```

```python
df['Total Time'] = df['Title'].apply(Extract_time)
```
